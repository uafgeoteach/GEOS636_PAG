```python
# Note that this is complicated for educational purposes as Python comes with a 
# function sum() that does what we implement below.

DEBUG = False

def print_debug(to_print):
    if(DEBUG):
        print("   DEBUG: %s" % to_print)

def sum_of_elements(in_array):
    print_debug("in sum_of_elements")
    
    s=0
    for x in in_array:
        print_debug("%g = %g + %g" % (s+x, s, x))
        s = s + x
        print_debug("s = %s" % (s) )    

    
    #print("end sum_of_elements")
    return s

#define list that we want to operate on
my_list = [1,23,34]

x = sum_of_elements(my_list)

print("----------------")
print("RESULT of our function:      %g" % x)

#this should be the same as this:
print("RESULT of built-in function: %g" % sum(my_list) )
```
